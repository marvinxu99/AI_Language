# Minimal BERT text-classification demo
Here’s a small example that classifies sentences as positive or negative sentiment using Hugging Face Transformers.

```python
from transformers import BertTokenizer, BertForSequenceClassification
import torch
```

```python
# 1️⃣ Load pretrained tokenizer & model
model_name = "bert-base-uncased"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name, num_labels=2)
```

```python
# 2️⃣ Example text
texts = [
    "I love this movie, it was fantastic!",
    "This film was boring and too long."
]
```

```python
# 3️⃣ Tokenize
inputs = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")
```

```python
# 4️⃣ Forward pass
with torch.no_grad():
    outputs = model(**inputs)
    probs = torch.softmax(outputs.logits, dim=1)
```

```python
# 5️⃣ Interpret results
labels = ["Negative", "Positive"]
for text, prob in zip(texts, probs):
    pred = labels[prob.argmax()]
    conf = prob.max().item() * 100
    print(f"{text}\n → {pred} ({conf:.2f}% confidence)\n")
```
